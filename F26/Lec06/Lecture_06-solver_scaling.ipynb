# Example of Solving Linear Systems of Different Sizes

```python
#import packages for example
import numpy as np
from timeit import timeit
import matplotlib.pyplot as plt
```

## Setting up and solving a linear system
Set linear system size

```python
N=10;
```

Setup random dense linear system with reference solution

```python
A = np.random.rand(N,N);
xref = np.ones(N);
b = np.dot(A,xref);
```

Solve the linear system by some method

```python
x = np.linalg.solve(A,b)
```

## Verify the result

```python
np.linalg.norm(x-xref,2)
```

## Measure time for solution

```python
t = timeit(lambda: np.linalg.solve(A,b), number=10)
print(f"{t} ms".format(t * 1000/ 10))
```

## Investigate scaling of solve time vs system size

```python
systemSize=[];
solnTime=[];
for i in range(4):
    N = 10**(i+1)
    A = np.random.rand(N,N);
    xref = np.ones(N);
    b = np.dot(A,xref);

    t = timeit(lambda: np.linalg.solve(A,b), number=10)
    
    systemSize.append(N)
    solnTime.append(t*1000/10)
    
plt.plot(systemSize,solnTime,'bx--')
plt.xlabel("System Size");
plt.ylabel("Solution Time (ms)")
```
