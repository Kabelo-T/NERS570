## Problem Decription

Two problems are used in this notebook:

1. The first is a Poisson problem written as:

$$-\nabla^2 u=g$$

In 1D, the first order cell-centered finite difference discretization is:
$$-\frac{1}{h^2}(u_{i-1}-2u_{i}+u_{i+1})=g$$

In 2D, the first order cell-centered finite difference discretization is:
$$
-\frac{u_{i+1,j}+u_{i-1,j}+u_{i,j+1}+u_{i,j-1}-4u_i}{h^2}=g
$$


In this notebook, we will show the iterative performance for the following stationary methods:
1. Jacobi Iterations (JI)
2. Gauss-Siedel Iterations (GS)
3. Algebric Multigrid Method (AMG)


 

## Jacobi and Gauss-Seidel for the 1D Poisson Problem
### 1D Poisson Problem

Let's begin with the 1D Poisson problem, thanks to other developers, we can easily construct the linear system for 1D poisson problem by:

```python
A = pyamg.gallery.poisson((100,), format='csr')

#You can examine the matrix structure here.
#Note it should be Tridiagonal with 2 on the diagonal and -1 on the sub/super-diagonal

print(A[0,0])
print(A)
```

The linear operator constructed is stored in compress sparse row format. To show convergence of each solver, we define the solution to be $0$. It follows then that $\bf{b}$ is also zero. Therefore, with an initial guess of a non-zero $\bf{x_0}$, the estimate of the solution for each iterative step is the true error.

Traditionally, $b$ is often used as the initial guess, but here we will use a random solution vector. Otherwise the solvers will converge to 0 very fast.

### Convergence Analysis
Spectral Radius of Iteration Matrices

```python
# Get L+D+U=A
L = scipy.sparse.tril(A,-1);
D = scipy.sparse.diags(A.diagonal());
U = scipy.sparse.triu(A,1);


F_j = np.linalg.inv(D.toarray())*(L+U);           #D^-1 * (L+U)
F_g = np.linalg.inv(D.toarray()+L.toarray())*(U); #(D+L)^-1 U

rho_j = np.max(np.absolute(np.linalg.eigvals(F_j))) #Converge at the rate of the largest eigenvalue (spectral radius)
rho_g = np.max(np.absolute(np.linalg.eigvals(F_g)))
print(rho_j)
print(rho_g)
```

### Low Wave Numbers (Flat Error Modes)

```python
k = 1 #wave number
x_0 = np.sin(np.arange(0,100)/100*np.pi*k) 
refx = np.zeros(A.shape[0])
b=A.dot(np.array(refx))
nSize=A.shape[0]
```

```python
plt.plot(x_0)
plt.show()
```

```python
%matplotlib notebook
from ipywidgets import *
import numpy as np
import IPython
import matplotlib.pyplot as plt
import scipy
from IPython.display import clear_output

nSize=A.shape[0]

jac_sol = nelin.solutionvector(nSize,x_0)
x_jac=  nelin.Jac(A,b,x0=x_0, tol=1e-9,maxiter=1000,callback=jac_sol.store)
resv_jac,resnorm_jac=jac_sol.getres(A,b)
 

gs_sol = nelin.solutionvector(nSize,x_0)
x_gs =  nelin.GS(A,b,x0=x_0, tol=1e-9,maxiter=1000,w=1,callback=gs_sol.store)
resv_gs,resnorm_gs=gs_sol.getres(A,b)

jaciter,jacsol=jac_sol.getsol()
gsiter,gssol=gs_sol.getsol()


# Create Sub plots
fig,axs = plt.subplots(2, 2,figsize=(1600./200,800./200), dpi=100)
fig.subplots_adjust(wspace=0.3)
axs[0][0].set_title('Jacobi Iteration Error Residual', fontsize=10)
axs[1][0].set_title('Gauss-Seidel Iteration Error Residual', fontsize=10)
axs[0][0].set_yscale('log')
axs[1][0].set_yscale('log')
axs[0][1].set_title('Jacobi Iteration Error', fontsize=10)
axs[1][1].set_title('GS Iteration Error', fontsize=10)
axs[0][1].set_yscale('log')
axs[1][1].set_yscale('log')
axs[0][0].set_xlim(0,jaciter)
axs[1][0].set_xlim(0,gsiter)
axs[0][0].set_ylim(1e-12,1)
axs[1][0].set_ylim(1e-12,1)
axs[0][1].set_xlim(0,nSize-1)
axs[1][1].set_xlim(0,nSize-1)
axs[0][1].set_ylim(1e-16,1)
axs[1][1].set_ylim(1e-14,1)

# animation function.  This is called sequentially
jacplot,= axs[0][0].plot([], [], color=colors[0])
gsplot,= axs[1][0].plot([], [], color=colors[1])
jacplote,= axs[0][1].plot([], [], color=colors[0])
gsplote,= axs[1][1].plot([], [], color=colors[1])


def update(w=40):
    ijac=w;
    igs=w;

    print('Start')
    if(w>=jaciter):
        ijac=jaciter-1
    if(w>=gsiter):
        igs=gsiter-1


    jacplot.set_data(np.arange(0, ijac+1), resnorm_jac[:ijac+1])
    gsplot.set_data(np.arange(0, igs+1), resnorm_gs[:igs+1])
   
    if(ijac>0):
        rhoj = resnorm_jac[ijac]/resnorm_jac[ijac-1];
    else:
        rhoj = 1.0;
    
    if(igs>0):
        rhog = resnorm_gs[igs]/resnorm_gs[igs-1];
    else:
        rhog = 1.0;

    axs[0][0].set_title('Jacobi Iteration Error Residual (rho=%1.5f)' % rhoj , fontsize=10)
    axs[1][0].set_title('Gauss-Seidel Iteration Error Residual (rho=%1.5f)'  % rhog, fontsize=10)
    
    jacplote.set_data(np.arange(0, nSize), np.abs(resv_jac[ijac,:]))
    gsplote.set_data(np.arange(0, nSize), np.abs(resv_gs[igs,:]))

    fig.suptitle('Iteration {}'.format(w), fontsize=20)
    fig.canvas.draw_idle()

interact(update,w=IntSlider(value=0,min=0,max=1000,step=50));

# If you cannot use iteractive mode just uncomment the nextwo lines    
# ani = animation.FuncAnimation(fig, update, repeat_delay=50,interval=50,repeat=False,frames=len(resnorm_gs)-1, blit=True)
# plt.show()
```

```python
plt.close() #Stop interacting with the plot
```

## Multigrid

```python
k = 1 #wave number
x_0 = np.sin(np.arange(0,100)/100*np.pi*k) 
refx = np.zeros(A.shape[0])
b=A.dot(np.array(refx))
nSize=A.shape[0]
```

```python
plt.plot(x_0)
plt.show()
```

```python
print(pyamg.ruge_stuben_solver(A,max_coarse=10))
```

```python
%matplotlib notebook
from ipywidgets import *
import numpy as np
import IPython
import matplotlib.pyplot as plt
import scipy
from IPython.display import clear_output

nSize=A.shape[0]

mg_sol = nelin.solutionvector(nSize,x_0)
x_mg=  nelin.MG(A,b,x0=x_0, tol=1e-9,maxiter=7,callback=mg_sol.store)
resv_mg,resnorm_mg=mg_sol.getres(A,b)

mgiter,mgsol=mg_sol.getsol()

# Create Sub plots
fig,axs = plt.subplots(1, 2,figsize=(1600./200,1500./200), dpi=100)
fig.subplots_adjust(wspace=0.3)
axs[0].set_title('MG Iteration Error Residual', fontsize=10)
axs[0].set_yscale('log')
axs[1].set_title('MG Iteration Error', fontsize=10)
axs[1].set_yscale('log')
axs[0].set_xlim(0,mgiter)
axs[0].set_ylim(1e-12,1)
axs[1].set_xlim(0,nSize-1)
axs[1].set_ylim(1e-16,1)


# animation function.  This is called sequentially
mgplot,= axs[0].plot([], [], color=colors[0])
mgplote,= axs[1].plot([], [], color=colors[0])

def update(w=40):
    img=w;

    print('Start')
    if(w>=mgiter):
        img=mgiter-1

    mgplot.set_data(np.arange(0, img+1), resnorm_mg[:img+1])
   
    mgplote.set_data(np.arange(0, nSize), np.abs(resv_mg[img,:]))

    fig.suptitle('Iteration {}'.format(w), fontsize=20)
    fig.canvas.draw_idle()

interact(update,w=IntSlider(value=0,min=0,max=7,step=1));
```
